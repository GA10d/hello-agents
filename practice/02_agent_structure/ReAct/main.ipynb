```python
import ReAct_Agent
import LLM_Client
import Tool_Search
from dotenv import load_dotenv
```

```python
llm_client = LLM_Client.HelloAgentsLLM()
tool_executor = Tool_Search.ToolExecutor()

agent_client = ReAct_Agent.ReActAgent(llm_client, tool_executor)
```

```python
search_description = "一个网页搜索引擎。当你需要回答关于时事、事实以及在你的知识库中找不到的信息时，应使用此工具。"
tool_executor.registerTool("Search", search_description, Tool_Search.search)
```

```python
agent_client.run("帮我上网查一下最新iphone折叠手机多少钱")
```
